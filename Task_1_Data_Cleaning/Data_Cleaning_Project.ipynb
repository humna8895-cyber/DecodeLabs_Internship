```python
import pandas as pd

df = pd.read_excel("raw_data.xlsx")

df.head()
```

```python
df.isnull().sum()
```

```python
median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)

print("Median Age:", median_age)
print("Missing Age after cleaning:", df["Age"].isnull().sum())
```

```python
median_quantity = df["Quantity"].median()

df["Quantity"] = df["Quantity"].fillna(median_quantity)

print("Median Quantity:", median_quantity)
print("Missing Quantity after cleaning:", df["Quantity"].isnull().sum())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(keep=False)]
```

```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python
print("City values before cleaning:")
print(df["City"].unique())
```

```python
df["City"] = df["City"].str.strip().str.title()

print("City values after cleaning:")
print(df["City"].unique())
```

```python
df["Product"] = df["Product"].str.strip().str.title()

print("Product values after cleaning:")
print(df["Product"].unique())
```

```python
print(df["Order_Date"].dtype)
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], dayfirst=True)

print(df["Order_Date"].dtype)
```

```python
print(df["Order_Date"].dt.strftime("%Y-%m-%d").head(10))
```

```python
print("Total records:", len(df))
print("Duplicate Order IDs:", df["Order_ID"].duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
print("Invalid dates:", df["Order_Date"].isnull().sum())
```

```python
print(df.dtypes)
```

```python
df["Order_Value"] = df["Order_Value"].round(2)
df["Quantity"] = df["Quantity"].round(2)

print(df[["Quantity", "Order_Value"]].head())
```

```python
df.to_excel("cleaned_data.xlsx", index=False)

print("Cleaned dataset saved successfully as cleaned_data.xlsx")
```
